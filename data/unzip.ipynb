```python
from pathlib import Path
file_dir = Path("raw/现场流程相关文件/")
target_extensions = {'.pdf', '.doc', '.docx', '.xlsx'}
file_paths = [
    p for p in file_dir.glob("*")
    if p.is_file() and p.suffix in target_extensions
]
```

```python
import subprocess
from pathlib import Path
raw_path = Path("现场流程相关文件.zip")
staging_path = Path("raw/")
staging_path.mkdir(parents=True, exist_ok=True)
# 使用系统 unzip 命令解决编码问题
try:
    # Linux/macOS
    subprocess.run(["unzip", "-O", "GBK", raw_path, "-d", str(staging_path)], check=True)
except FileNotFoundError:
    print("解压失败")
```
